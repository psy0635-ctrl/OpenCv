```python
# \n을 이용해 print()를 한 번만 사용할 수 있어요!
print("안녕하세요!")
print("반갑습니다.")

print("안녕하세요!\n반갑습니다.")

print("\t\"py\nthon5\"")
```

```python
# Python의 print()는 자동으로 \n, 줄바꿈을 합니다.
# end를 이용해 이를 변경할 수 있어요!
# 여러번 줄바꿈은 \n\n (두 줄 바꿈)과 같이 이용하면 됩니다.
print("Hello")
print("Python!")

print("Hello", end=" ")
print("Python!")

print("Hello", end="😊")
print("Python!")
print("---")
# 단, 다음 print에 적용되지는 않는다!
print("Hello")
```

```python
# Python의 print()는 , 를 이용할 때 자동으로 공백을 삽입합니다.
# sep를 이용해 이를 변경할 수 있어요!
print("2026", "03", "04", sep="-")
print("2026", "03", "04", sep="*")
```

```python
# Python 과제 2

# 1. 자기소개 프로그램 사용자로부터 이름, 나이, 취미를 입력받아 출력하는 프로그램을 작성하세요.

name = input("이름을 입력하세요 : ")
age = input ("나이를 입력하세요 : ")
hobby = input("취미를 입력하세요 : ")

print("\n[자기소개]")
print("이름 : ", name)
print("나이 : ", age)
print("취미 : ",hobby)
```

```python
# 2. 두 숫자를 입력받아, 서로 나눈 나머지를 출력하는 프로그램을 작성하세요.

num1 = int(input("첫번째 숫자를 입력하세요 : "))
num2 = int(input("두번째 숫자를 입력하세요 : "))

print("두 숫자를 나눈 나머지 : ",num1 % num2)
```

```python
# 3. 두 점수를 입력받아 평균을 구해 출력하는 프로그램을 작성하세요.

num1 = float(input("첫번째 점수를 입력하세요 : "))
num2 = float(input("두번째 점수를 입력하세요 : "))

print("두 점수의 평균 : ",(num1 + num2) / 2)
```
