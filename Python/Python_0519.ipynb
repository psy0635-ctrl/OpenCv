# 3. 자료형 (Data Type)

## 자료형이란?

- 변수에 저장되는 **데이터의 종류**
- Python은 변수에 값을 할당하면 자료형이 **자동으로 결정됨** (동적 타이핑)
- 자료형에 따라 **사용할 수 있는 연산이 달라짐**

## Python의 기본 자료형 4가지

| 자료형 | 이름 | 예시 | 설명 |
|--------|------|------|------|
| `int` | 정수형 | `25`, `-3`, `0` | 소수점이 없는 숫자 |
| `float` | 실수형 | `175.5`, `3.14`, `-0.1` | 소수점이 있는 숫자 |
| `str` | 문자열 | `"차가을"`, `'Hello'` | 따옴표로 감싼 글자 |
| `bool` | 논리형 | `True`, `False` | 참/거짓 두 가지 값 |

````python
# 각 자료형 변수에 값 저장
age = 25              # 정수 (int)
height = 175.5        # 실수 (float)
name = "박수용"        # 문자열 (str)
is_student = True     # 논리 (bool)

print(age)
print(height)
print(name)
print(is_student)
````

# 3. 리스트 `list`

## 리스트란?

- **여러 값을 순서대로 담는** 자료형
- 대괄호 `[ ]` 로 생성, 쉼표로 구분
- 인덱스로 접근 (0부터 시작 — C 배열과 동일)

## C 배열 vs Python list

| 항목 | C 배열 | Python list |
|---|---|---|
| 선언 | `int arr[5];` (크기 고정) | `arr = []` (크기 자유) |
| 타입 | 한 가지 타입만 | **여러 타입 혼합 가능** |
| 크기 변경 | 불가 | **`append()`, `pop()` 등으로 자유롭게** |
| 슬라이싱 | 없음 (반복문 직접 작성) | `arr[1:4]` 한 줄로 끝 |
| 길이 확인 | `sizeof(arr)/sizeof(arr[0])` | `len(arr)` |

````python
# 리스트 생성
scores = [85, 90, 78, 92, 88]
print(scores)
print(type(scores))
print("길이:", len(scores))

# 여러 타입 혼합도 OK (C 배열에선 불가능!)
mixed = [1, 3.14, "hello", True, [1, 2]]
print(mixed)
````

## 인덱싱과 슬라이싱

- 인덱스는 **0부터** 시작 (C와 동일)
- **음수 인덱스**: 뒤에서부터 (Python의 특별한 기능!)
- **슬라이싱** `[시작:끝:간격]` — 끝 인덱스는 **포함하지 않음**

````python
scores = [85, 90, 78, 92, 88]
#         0   1   2   3   4
#        -5  -4  -3  -2  -1

# 인덱싱
print(scores[0])    # 85 (첫 번째)
print(scores[-1])   # 88 (마지막)
print(scores[-2])   # 92 (뒤에서 두 번째)

# 슬라이싱
print(scores[1:4])   # [90, 78, 92]  (인덱스 1~3)
print(scores[:3])    # [85, 90, 78]  (처음부터 인덱스 2까지)
print(scores[2:])    # [78, 92, 88]  (인덱스 2부터 끝까지)
print(scores[::2])   # [85, 78, 88]  (2칸씩 건너뛰며)
print(scores[::-1])  # [88, 92, 78, 90, 85]  (뒤집기)
````

## 값 변경 (가변, mutable)

````python
scores = [85, 90, 78, 92, 88]

# 특정 위치 값 변경
scores[0] = 100
print(scores)   # [100, 90, 78, 92, 88]

# 슬라이스로 한꺼번에 변경
scores[1:3] = [99, 99]
print(scores)   # [100, 99, 99, 92, 88
````

## 자주 쓰는 리스트 메서드

| 메서드 | 기능 |
|---|---|
| `append(x)` | 끝에 x 추가 |
| `insert(i, x)` | 인덱스 i 위치에 x 삽입 |
| `remove(x)` | x 값을 가진 **첫 번째** 요소 삭제 |
| `pop(i)` | 인덱스 i의 값을 꺼내며 삭제 (생략 시 마지막) |
| `sort()` | 오름차순 정렬 (원본 변경) |
| `reverse()` | 순서 뒤집기 (원본 변경) |
| `count(x)` | x의 개수 |
| `index(x)` | x의 첫 인덱스 |

````python
nums = [3, 1, 4, 1, 5, 9, 2, 6]

nums.sort()
print(nums)   # [1, 1, 2, 3, 4, 5, 6, 9]

nums.reverse()
print(nums)   # [9, 6, 5, 4, 3, 2, 1, 1]

print("1의 개수:", nums.count(1))
print("5의 위치:", nums.index(5))
````

````python
## 학생 점수 처리

# 1. 리스트 생성
scores = [85, 90, 78, 92, 88]

# 2. 점수 추가
scores.append(75)
print("추가 후:", scores)

# 3. 최저점 제거
scores.remove(min(scores))
print("최저점 제거 후:", scores)

# 4. 평균
average = sum(scores) / len(scores)
print("평균:", average)
````

# 4. 튜플 `tuple`

## 튜플이란?

- list와 거의 비슷하지만 **한번 만들면 변경 불가** (immutable)
- 소괄호 `( )` 로 생성
- "이 데이터는 절대 바뀌면 안 돼!"

## list vs tuple

| | list | tuple |
|---|---|---|
| 기호 | `[ ]` | `( )` |
| 변경 가능? | 가능 | 불가능 |
| 사용 예 | 학생 점수, 장바구니 | 좌표 (x, y), 색상 (R,G,B) |

````python
# 튜플 생성
point = (3, 5)
color = (255, 100, 50)

print(point)
print(type(point))

# 인덱싱, 슬라이싱은 list와 동일
print(point[0], point[1])
print(color[0:2])

# 하지만 변경은 불가!
# point[0] = 10   # TypeError 발생
````

# 6. 집합 `set`

## set이란?

- **중복을 허용하지 않는** 값의 모음
- **순서 없음** (인덱싱 불가)
- 중괄호 `{ }` 로 생성 (단, 빈 set은 `set()` 으로)
- 수학의 집합과 동일한 개념: 합집합, 교집합, 차집합 가능

## 언제 쓰는가?

- 리스트에서 **중복 제거**할 때
- 두 그룹의 **공통 요소** 찾을 때

````python
# 중복 제거
nums = [1, 2, 2, 3, 3, 3, 4]
unique = set(nums)
print(unique)              # {1, 2, 3, 4}
print(list(unique))        # 다시 리스트로

# 수학의 집합 연산
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(A | B)   # 합집합: {1, 2, 3, 4, 5, 6}
print(A & B)   # 교집합: {3, 4}
print(A - B)   # 차집합: {1, 2}
````

````python
## 거스름돈 계산기 만들어 보기
## -> 다음 프로그램을 IPO 구조로 작성하세요.

# 입력
price = int(input("물건 가격 : "))
paid = int(input("낸 돈 : "))

# 처리
change = paid - price   # 거스름돈 = 낸 돈 - 물건 가격

# 5만원권 계산
w50000 = change // 50000   # 거스름돈을 50000으로 나눈 몫 → 5만원권 장수
change = change %  50000   # 5만원권 주고 남은 돈

# 1만원권 계산
w10000 = change // 10000   # 남은 돈을 10000으로 나눈 몫 → 1만원권 장수
change = change %  10000   # 1만원권 주고 남은 돈

# 5천원권 계산
w5000  = change // 5000    # 남은 돈을 5000으로 나눈 몫 → 5천원권 장수
change = change %  5000    # 5천원권 주고 남은 돈

# 1천원권 계산
w1000  = change // 1000    # 남은 돈을 1000으로 나눈 몫 → 1천원권 장수

# 출력
print("거스름돈: " , paid - price, "원")
print("5만원권 :" , w50000 ,"장")
print("1만원권 :" , w10000 ,"장")
print("5천원권 :" , w5000 ,"장")
print("1천원권 :" , w1000 ,"장")
````
