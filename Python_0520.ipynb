# 2. 비교 연산자

## 비교 연산자란?

두 값을 비교하여 **`True` 또는 `False`** 를 반환하는 연산자.
**결과는 항상 `bool` 자료형**입니다.

| 연산자 | 의미 | 예시 |
|---|---|---|
| `==` | 같다 | `5 == 5` → True |
| `!=` | 다르다 | `5 != 3` → True |
| `>` | 크다 | `5 > 3` → True |
| `<` | 작다 | `5 < 3` → False |
| `>=` | 크거나 같다 | `5 >= 5` → True |
| `<=` | 작거나 같다 | `5 <= 3` → False |


````python
a = 10
b = 20

print(a == b)   # False
print(a != b)   # True
print(a < b)    # True
print(a >= 10)  # True

# 결과는 항상 bool
result = (a < b)
print(result, type(result))
````

# 3. 논리 연산자

## 논리 연산자란?

여러 조건을 **결합**하는 연산자.

| Python | C | 의미 |
|---|---|---|
| `and` | `&&` | 둘 다 참일 때 True |
| `or` | `\|\|` | 하나라도 참이면 True |
| `not` | `!` | 참/거짓 뒤집기 |

> **Python은 영어 단어를 그대로 씁니다.**

````python
age = 22
score = 85

# and: 둘 다 참
print((age >= 20) and (score >= 80))   # True

# or: 하나라도 참
print(age < 20 or score >= 80)     # True (score 조건이 참)

# not: 뒤집기
print(not (age >= 20))             # False
````

# 4. 조건문 `if`

## 기본 구조

```python
if 조건:
    실행할_코드
```

> **C와의 핵심 차이점**
> - 조건 뒤에 **콜론 `:`**
> - 중괄호 `{ }` 대신 **들여쓰기**로 블록 표시 (관례: tap (space 4칸))
> - 들여쓰기가 틀리면 오류 (꼭 기억하기)

````python
score = 85

if score >= 60:
    print("합격!")
    print("축하합니다.")

print("프로그램 종료")   # 들여쓰기 X → if 밖
````

## if - else

조건이 거짓일 때 실행할 코드가 있다면 `else` 추가.

````python
score = 55

if score >= 60:
    print("합격!")
else:
    print("불합격...")
````

## if - elif - else

조건이 **여러 개**일 때 `elif` (else if의 줄임) 사용.

````python
score = 75

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
elif score >= 60:
    print("D")
else:
    print("F")
````

## 중첩 조건문

`if` 안에 또 `if` 를 넣을 수 있습니다.

````python
age = 22
has_ticket = True

if age >= 18:
    if has_ticket:
        print("입장 가능")
    else:
        print("티켓을 구매하세요")
else:
    print("미성년자 입장 불가")
````

````python
# 입력
height_cm = float(input("키(cm): "))
weight_kg = float(input("몸무게(kg): "))

# 처리
height_m = height_cm / 100
bmi = weight_kg / (height_m ** 2)

# 출력
print("BMI:", bmi)

if bmi < 18.5:
    print("저체중")
elif bmi < 23:
    print("정상")
elif bmi < 25:
    print("과체중")
else:
    print("비만")
````

# 5. 반복문 `while`

## 기본 구조

```python
while 조건:
    반복할_코드
```

**조건이 참인 동안** 계속 반복.
조건이 거짓이 되면 빠져나옴.

> 콜론과 들여쓰기 주의.

````python
# 1부터 5까지 출력
i = 1
while i <= 5:
    print(i)
    i += 1   # i = i + 1 (C와 동일)

print("끝!")
````

> **무한 루프 주의**
>
> 반복 변수를 갱신하지 않으면 **무한히 반복**됩니다.
> Colab에서는 셀 왼쪽 ⏹ 버튼으로 중단할 수 있어요.
>
> ```python
> i = 1
> while i <= 5:
>     print(i)
>     # i += 1
> ```

## break 와 continue

| 키워드 | 역할 |
|---|---|
| `break` | 반복문 **즉시 종료** |
| `continue` | 현재 반복만 **건너뛰고** 다음으로 |


````python
# break: 5에서 멈추기
i = 1
while i <= 10:
    if i == 5:
        break
    print(i, end=" ")
    i += 1
# 출력: 1 2 3 4
````

````python
# continue: 짝수만 건너뛰기
i = 0
while i < 10:
    i += 1
    if i % 2 == 0:
        continue
    print(i, end=" ")
# 출력: 1 3 5 7 9
````

````python
# while문을 사용하여 1~100의 3의 배수를 제외한 합을 구하세요.

total = 0  # 합계를 저장할 변수 (처음엔 0)
i = 1      # 1부터 시작

while i <= 100:        # i가 100 이하인 동안 반복
    if i % 3 == 0:     # i가 3의 배수라면?
        i += 1         # 다음 숫자로 넘어가고
        continue       # 아래 코드 무시하고 while로 돌아감

    total += i  # 3의 배수가 아니면 total에 더함
    i += 1      # 다음 숫자로 이동

print("1~100의 3의 배수를 제외한 합:", total)  # 결과 출력

# i = 1  → 3의 배수 아님 → total에 1 더함
# i = 2  → 3의 배수 아님 → total에 2 더함
# i = 3  → 3의 배수!    → 건너뜀 ✋
# i = 4  → 3의 배수 아님 → total에 4 더함
# i = 5  → 3의 배수 아님 → total에 5 더함
# i = 6  → 3의 배수!    → 건너뜀 ✋
# ...
# i = 101 → while 조건 실패 → 종료
````

````python
# 비밀번호가 "1234" 일 때, 사용자가 맞출 때까지 반복해서 입력받는 프로그램을 작성하세요.

password = "1234"

while True:
    user_input = input("비밀번호 입력: ")
    if user_input == password:
        print("로그인 성공!")
        break
    else:
        print("틀렸습니다. 다시 입력하세요.")
````

# 6. 반복문 `for`

## Python의 for는 다르다

C의 for는 인덱스 변수를 직접 관리:
```c
for (int i = 0; i < 5; i++) { ... }
```

Python의 for는 **컨테이너의 요소를 하나씩 꺼냄**:
```python
for x in 컨테이너:
    실행할_코드
```


````python
# 리스트 순회
fruits = ["사과", "바나나", "포도"]

for fruit in fruits:
    print(fruit)
````

````python
# 문자열도 순회 가능 (한 글자씩)
for c in "Python":
    print(c)
````

## `range()` 함수

특정 횟수만큼 반복하고 싶을 때 사용.

| 형태 | 의미 |
|---|---|
| `range(n)` | 0부터 n-1까지 |
| `range(a, b)` | a부터 b-1까지 |
| `range(a, b, step)` | a부터 b-1까지, step 간격 |

````python
# 0부터 4까지
for i in range(5):
    print(i, end=" ")
print()

# 1부터 5까지
for i in range(1, 6):
    print(i, end=" ")
print()

# 1부터 10까지 2씩 (홀수)
for i in range(1, 11, 2):
    print(i, end=" ")
print()

# 10부터 1까지
for i in range(10, 0, -1):
    print(i, end=" ")
````

> **C의 `for(i=0; i<n; i++)` -> Python에서 `for i in range(n)`** 입니다.

````python
# list에서 80점 넘은 학생만 골라내려면?

scores = [85, 72, 90, 65, 88, 95, 58, 77]

# 80점 이상만 출력
for s in scores:
    if s >= 80:
        print(s, "점 - 우수")
````

````python
# 구구단 - for문 두 개를 중첩해서 2단부터 9단까지 출력하세요.

for dan in range(2, 10):
    for i in range(1, 10):
        print(dan, "x", i, "=", dan * i)
    print("---")
````

````python
# 숫자 맞히기 게임
# 다음 사양으로 게임을 만드세요.

# 게임 규칙
# 컴퓨터가 1~100 사이의 랜덤 숫자를 정함
# 사용자가 숫자를 입력
# 입력한 숫자가 정답보다 크면 "더 작은 수 입니다.", 작으면 "더 큰 수 입니다." 출력
# 정답을 맞히면 "정답입니다! N번 만에 맞췄습니다." 출력 후 종료

import random

answer = random.randint(1, 100)  #  1 ~ 100 사이 랜덤 정수
count = 0

while True:
  num = int(input("숫자 입력 : "))
  count += 1
  if num < answer:
    print("더 큰 수 입니다.")
  elif num > answer:
      print("더 작은 수 입니다.")
  else:
    print("정답입니다." , count , "번 만에 맞췄습니다.")
    break
````
