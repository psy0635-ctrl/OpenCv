## 함수 정의와 호출

### 기본 문법

```python
def 함수이름():
    실행할 코드
    실행할 코드
```

- `def` : **def**ine(정의)의 약자
- 콜론(`:`) 필수
- 함수 내부는 **들여쓰기(4칸)** 필수
- 정의만으로는 실행 안 됨 → **호출**해야 실행

````python
# 가장 간단한 함수 정의와 호출

def greet():               # 1) 함수 정의
    print("안녕하세요!")
    print("파이썬 함수 시작!")

# 여기까지는 아무것도 실행되지 않음

greet()    # 2) 함수 호출 -> 실행
greet()    # 3) 다시 호출
````

````python
# 주의: 함수는 정의 이후에 호출해야 한다

# say_hello()  ← 이 위치에서 호출하면 NameError

def say_hello():
    print("Hello!")

say_hello()
````

## 반환값 (return)

### return의 직관

지금까지의 함수는 `print`로 화면에 출력만 했어요.  
하지만 함수가 **계산한 결과를 돌려줄 수 없다면** 다른 곳에서 쓸 수 없습니다.

```
자판기 버튼 누름(호출) → [자판기 내부 동작] → 음료수 나옴(return)
```

```python
def 함수이름():
    계산
    return 결과값    # ← 이 값이 호출한 곳으로 전달됨
```

````python
# return이 없는 함수 vs return이 있는 함수

# print만 하는 함수: 결과를 저장할 수 없음
def add_no_return(a, b):
    print(a + b)

result1 = add_no_return(3, 5)
print("저장된 값:", result1)   # None 출력됨!

# return 있는 함수: 결과를 변수에 저장 가능
def add(a, b):
    return a + b

result2 = add(3, 5)
print("저장된 값:", result2)   # 8 출력!
print("두 배:", result2 * 2)   # 다른 연산에도 활용 가능
````

````python
# return은 함수를 즉시 종료시킵니다.

def check_brightness(value):
    if value < 0 or value > 255:
        return "잘못된 값"   # 여기서 함수 종료
    if value < 85:
        return "어두움"
    elif value < 170:
        return "보통"
    else:
        return "밝음"

print(check_brightness(50))   # 어두움
print(check_brightness(128))  # 보통
print(check_brightness(220))  # 밝음
print(check_brightness(300))  # 잘못된 값
````

````python
# 여러 값 반환하기

def get_stats(pixels):
    """픽셀 리스트의 최솟값, 최댓값, 평균을 반환"""
    min_val = min(pixels)
    max_val = max(pixels)
    avg_val = sum(pixels) / len(pixels)
    return min_val, max_val, avg_val   # 세 값을 한꺼번에 반환

pixels = [45, 120, 200, 89, 156, 33, 178]

# 방법 1: 개별 변수로 받기
lo, hi, avg = get_stats(pixels)
print(f"최솟값: {lo}, 최댓값: {hi}, 평균: {avg:.1f}")

# 방법 2: 튜플로 받기
stats = get_stats(pixels)
print("전체:", stats)
````

## 매개변수 (Parameter)

### 용어 정리

| 용어 | 설명 | 예시 |
|------|------|------|
| **매개변수** (parameter) | 함수 정의 시 입력 이름 | `def f(x, y):` 에서 `x`, `y` |
| **인자** (argument) | 함수 호출 시 실제 전달하는 값 | `f(3, 5)` 에서 `3`, `5` |

```python
#      매개변수
#      ↓   ↓
def add(a,  b):
    return a + b

#     인자
#    ↓  ↓
add(10, 20)
```

### 위치 인자 (Positional Argument)

가장 기본적인 방식. **순서대로** 매개변수에 대입됩니다.

````python
def describe_pixel(row, col, value):
    print(f"위치 ({row}, {col})의 밝기: {value}")

describe_pixel(10, 20, 128)    # row=10, col=20, value=128
describe_pixel(5, 15, 200)     # row=5,  col=15, value=200

# 순서가 바뀌면 의미도 바뀝니다
describe_pixel(128, 10, 20)    # row=128, col=10, value=20  ← 의도와 다름
````

### 기본값 매개변수 (Default Parameter)

### 직관
자주 쓰는 값을 미리 지정해두면, 호출 시 생략할 수 있습니다.

```python
def 함수이름(필수값, 선택값=기본값):
```

> **규칙**: 기본값이 있는 매개변수는 **항상 뒤쪽**에 위치해야 함!

````python
# 영상 밝기 조절 함수 (기본값: alpha=1.0, beta=0)
# 실제 OpenCV의 convertScaleAbs와 유사한 구조

def adjust_brightness(pixel, alpha=1.0, beta=0):
    """
    pixel: 원본 픽셀값
    alpha: 대비 조절 (기본값 1.0 = 변화 없음)
    beta : 밝기 조절 (기본값 0  = 변화 없음)
    """
    result = alpha * pixel + beta
    return max(0, min(255, int(result)))   # 0~255 범위 제한

# 기본값만 사용 (변화 없음)
print(adjust_brightness(100))              # alpha=1.0, beta=0

# 밝기만 올리기 (beta만 변경)
print(adjust_brightness(100, beta=50))     # 100 + 50 = 150

# 대비 높이기 (alpha만 변경)
print(adjust_brightness(100, alpha=1.5))   # 100 * 1.5 = 150

# 둘 다 변경
print(adjust_brightness(100, alpha=1.5, beta=30))  # 180
````

````python
# 기본값 매개변수 위치 규칙 확인

# 오류 발생: 기본값 있는 것이 앞에 오면 안 됨
# def wrong(alpha=1.0, pixel):   # SyntaxError!
#     return alpha * pixel

# 올바른 순서: 기본값 없는 것 → 기본값 있는 것
def correct(pixel, alpha=1.0):
    return alpha * pixel

print(correct(100))         # alpha는 생략 → 1.0 사용
print(correct(100, 2.0))    # alpha = 2.0
````

키워드 인자 (Keyword Argument)

### 직관
인자를 전달할 때 **이름을 명시**하면 순서와 상관없이 원하는 매개변수에 전달됩니다.

```python
함수이름(매개변수이름=값)
```

````python
def draw_rectangle(x, y, width, height, color="black"):
    print(f"({x}, {y})에서 {width}×{height} 사각형, 색상: {color}")

# 방법 1: 위치 인자 (순서대로)
draw_rectangle(10, 20, 100, 50)

# 방법 2: 키워드 인자 (이름 지정, 순서 무관)
draw_rectangle(width=100, height=50, x=10, y=20)

# 방법 3: 혼합 (위치 먼저, 키워드 나중)
draw_rectangle(10, 20, height=50, width=100, color="red")
````

````python
# 키워드 인자의 장점: 코드 가독성이 높아진다

import cv2
import numpy as np

img = np.zeros((200, 300, 3), dtype=np.uint8)

# 키워드 없이 → 각 숫자가 무엇인지 알기 어려움
cv2.rectangle(img, (10, 20), (100, 80), (0, 255, 0), 2)

# 직접 만든 래퍼 함수에 키워드 인자 활용
def draw_box(image, x1, y1, x2, y2, color=(0, 255, 0), thickness=2):
    cv2.rectangle(image, (x1, y1), (x2, y2), color, thickness)
    return image

# 키워드 인자로 호출 → 각 값의 의미가 명확!
result = draw_box(img, x1=10, y1=20, x2=100, y2=80,
                  color=(255, 0, 0), thickness=3)
print("박스 그리기 완료!")
````

## 세 가지 방식 비교 정리

| 방식 | 문법 | 특징 |
|------|------|------|
| **위치 인자** | `f(10, 20)` | 순서대로 대입, 가장 기본 |
| **기본값 매개변수** | `def f(x, y=0):` | 생략 가능한 옵션 설정 |
| **키워드 인자** | `f(y=20, x=10)` | 이름 명시, 순서 무관 |

### 혼합 사용 순서 규칙

```
함수(위치인자, ..., 키워드인자=값, ...)
     ↑ 위치가 항상 먼저!  ↑ 키워드는 뒤에
```

````python
# 세 가지 방식을 모두 활용한 이미지 처리 함수

def process_image(image, method, alpha=1.0, beta=0, verbose=False):
    """
    image  : 처리할 이미지 (위치 인자, 필수)
    method : 처리 방법 문자열 (위치 인자, 필수)
    alpha  : 대비 계수 (기본값 1.0)
    beta   : 밝기 오프셋 (기본값 0)
    verbose: 상세 출력 여부 (기본값 False)
    """
    if verbose:
        print(f"[처리 시작] method={method}, alpha={alpha}, beta={beta}")

    # 실제 처리는 생략, 처리된 이미지 반환
    result = image.copy() if image is not None else None

    if verbose:
        print("[처리 완료]")
    return result

import numpy as np
img = np.zeros((100, 100), dtype=np.uint8)

# 필수 인자만
process_image(img, "brightness")

# 기본값 덮어쓰기
process_image(img, "contrast", alpha=1.5)

# 키워드 인자로 순서 변경 + verbose 켜기
process_image(img, "combined", verbose=True, beta=30, alpha=1.2)
````

함수 바로 아래에 `"""..."""` 형식으로 설명을 달면 나중에 `help()`로 확인 가능합니다.

````python
def normalize_pixel(value, min_val=0, max_val=255):
    """
    픽셀값을 0.0 ~ 1.0 범위로 정규화합니다.

    매개변수
        value   (int/float): 정규화할 픽셀값
        min_val (int): 최솟값 기준 (기본값: 0)
        max_val (int): 최댓값 기준 (기본값: 255)

    반환:
        float: 정규화된 값 (0.0 ~ 1.0)
    """
    return (value - min_val) / (max_val - min_val)

# help()로 docstring 확인
help(normalize_pixel)

print(normalize_pixel(0))     # 0.0
print(normalize_pixel(128))   # 0.502...
print(normalize_pixel(255))   # 1.0
````

### 실습 3. 응용 - 픽셀 판별기

픽셀값을 받아 밝기 단계를 반환하는 함수를 완성하세요.

| 픽셀값 범위 | 반환값 |
|-------------|--------|
| 0 ~ 63 | `"매우 어두움"` |
| 64 ~ 127 | `"어두움"` |
| 128 ~ 191 | `"밝음"` |
| 192 ~ 255 | `"매우 밝음"` |
| 그 외 | `"유효하지 않은 값"` |

````python
def classify_brightness(pixel):
    # 유효 범위 벗어난 값 먼저 처리 (음수 또는 255 초과)
    if pixel < 0 or pixel > 255:
        return "유효하지 않은 값"
    elif pixel <= 63:
        return "매우 어두움"
    elif pixel <= 127:
        return "어두움"
    elif pixel <= 191:
        return "밝음"
    else:
        return "매우 밝음"

# 테스트
for val in [0, 63, 64, 127, 128, 191, 192, 255, -1, 300]:
    print(f"픽셀값 {val:4d} → {classify_brightness(val)}")
````
